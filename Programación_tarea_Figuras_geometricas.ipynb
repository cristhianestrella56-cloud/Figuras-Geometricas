```python
import math


class FiguraGeometrica:
    def __init__(self, alto, ancho):
        self.__alto = None
        self.__ancho = None
        self.alto = alto
        self.ancho = ancho

    @property
    def alto(self):
        return self.__alto

    @alto.setter
    def alto(self, valor):
        if valor <= 0:
            raise ValueError("El alto debe ser positivo")
        self.__alto = valor

    @property
    def ancho(self):
        return self.__ancho

    @ancho.setter
    def ancho(self, valor):
        if valor <= 0:
            raise ValueError("El ancho debe ser positivo")
        self.__ancho = valor

    def area(self):
        return self.__alto * self.__ancho

    def __str__(self):
        return f'Alto={self.__alto}, Ancho={self.__ancho}'


class Color:
    def __init__(self, color):
        self.__color = None
        self.color = color

    @property
    def color(self):
        return self.__color

    @color.setter
    def color(self, valor):
        if not valor or not isinstance(valor, str):
            raise ValueError("El color debe ser un texto válido")
        self.__color = valor

    def __str__(self):
        return f'Color={self.__color}'


class Cuadrado(FiguraGeometrica, Color):
    def __init__(self, lado, color):
        FiguraGeometrica.__init__(self, lado, lado)
        Color.__init__(self, color)

    def area(self):
        return self.alto ** 2

    def __str__(self):
        return f'Cuadrado({FiguraGeometrica.__str__(self)}, {Color.__str__(self)}, Area={self.area():.2f})'


class Rectangulo(FiguraGeometrica, Color):
    def __init__(self, alto, ancho, color):
        FiguraGeometrica.__init__(self, alto, ancho)
        Color.__init__(self, color)

    def area(self):
        return self.alto * self.ancho

    def es_cuadrado(self):
        return self.alto == self.ancho

    def __str__(self):
        tipo = "Cuadrado" if self.es_cuadrado() else "Rectangulo"
        return f'{tipo}({FiguraGeometrica.__str__(self)}, {Color.__str__(self)}, Area={self.area():.2f})'


class Circunferencia(Color):
    def __init__(self, radio, color):
        super().__init__(color)
        self.__radio = None
        self.radio = radio

    @property
    def radio(self):
        return self.__radio

    @radio.setter
    def radio(self, valor):
        if valor <= 0:
            raise ValueError("El radio debe ser positivo")
        self.__radio = valor

    def area(self):
        return math.pi * (self.__radio ** 2)

    def diametro(self):
        return self.__radio * 2

    def __str__(self):
        return f'Circunferencia(Radio={self.__radio}, {Color.__str__(self)}, Area={self.area():.2f})'


def mostrar_detalle(figura):
    print("Tipo:", type(figura).__name__)
    print(figura)
    print("Area:", figura.area())
    print("-" * 40)


if __name__ == "__main__":
    f1 = Cuadrado(5, "Rojo")
    f2 = Rectangulo(4, 6, "Azul")
    f3 = Rectangulo(3, 3, "Amarillo")
    f4 = Circunferencia(3, "Verde")

    figuras = [f1, f2, f3, f4]

    for figura in figuras:
        mostrar_detalle(figura)

    print("Cambio de valores con setters")
    f1.alto = 7
    f1.ancho = 7
    f1.color = "Negro"

    f4.radio = 5
    f4.color = "Morado"

    for figura in figuras:
        mostrar_detalle(figura)

    print("Polimorfismo con lista dinámica")
    nuevas_figuras = [
        Cuadrado(2, "Blanco"),
        Rectangulo(10, 2, "Gris"),
        Circunferencia(1.5, "Celeste")
    ]

    total_area = 0
    for figura in nuevas_figuras:
        total_area += figura.area()
        print(figura)

    print("Area total:", round(total_area, 2))
```
